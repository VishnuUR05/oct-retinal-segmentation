# Prepare and visualize data #

```python
import os
from natsort import natsorted
import cv2
import numpy as np
import pathlib
import pandas as pd

rootdir = '../../AMD_DME_Normals/Detection/'

allImages = []
        
for subdir, dirs, files in os.walk(rootdir):
    for file in files:
        if(".png" in file or ".PNG" in file or ".TIFF" in file or ".TIF" in file or ".jpeg" in file):
            allImages.append(os.path.join(subdir, file))
            
allImages = natsorted(allImages)
```

# Dataframe for all bounding box annotations #

```python
df = pd.read_csv(rootdir+"all_bounding_boxes.csv")
```

# Visualize randomly selected example #

```python
#id_random = np.random.choice(range(0, len(allImages)), 1)[0]
id_random = 60;

df_single_image = df.loc[df['image'] == allImages[id_random].split("AMD_DME_Normals/Detection/Images")[-1][1:]] 

df_single_image
```

```python
import imageio
import cv2
from matplotlib import pylab as plt

boxes = [];
for i, row in df_single_image.iterrows():
    boxes.append([row['xmin'], row['ymin'], row['xmax'], row['ymax']])
```

```python
def show(im, boxes):
    for box in boxes:
        cv2.rectangle(im,
            (box[0], box[1]),
            (box[2], box[3]),
            (255,128,0), 1)
    return im

img_to_show = show(imageio.imread(allImages[id_random]), boxes)
```

# Plot example with bounding boxes #

```python
plt.figure(figsize=(10,10))
plt.imshow(img_to_show,cmap="gray")
```
